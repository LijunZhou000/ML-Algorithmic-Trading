```python
def get_data(specs, symbol):
    info = specs[symbol]
    ib = IB()
    ib.connect("127.0.0.1", 7497, clientId=2)
    future = Future(symbol, "202606", info['exchange'], info['currency'])
    data = ib.reqHistoricalData(info['conId'], future, "", "6 M", "1 min", "TRADES", 1, 1, False, [])
    ib.disconnect()
    return data
```
