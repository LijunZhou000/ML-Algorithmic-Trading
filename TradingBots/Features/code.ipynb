```python
from tradepy.load import load_future
from tradepy.features import generate_features, resample_ohlcv, daily_ohlcv_cummulative, prepare_accumulated_ohlcv, merge_intraday_daily
```

```python
df_5_min = resample_ohlcv(df_min)
```

```python
df_5_min_features = generate_features(df_5_min)
```

```python
df_dia_cum = daily_ohlcv_cummulative(df_5_min)
```

```python
df_dia_cum_features = prepare_accumulated_ohlcv(df_dia_cum)
```

```python
df_final = merge_intraday_daily(df_dia_cum_features, df_5_min_features)
```
