```python
df_min_bp = load_future('bp1', 'min')[-1]
df_min_cl = load_future('cl1', 'min')[-1]
df_min_gc = load_future('gc1', 'min')[-1]
df_min_es = load_future('es1', 'min')[-1]
```

```python
for nombre, df in [('bp1', df_min_bp), ('cl1', df_min_cl), ('gc1', df_min_gc), ('es1', df_min_es)]:
    mask = df['high'] == df['low']
    total = len(df)
    pct = mask.sum() / total * 100
    # distribución de volumen en esas filas
    vol_dist = df.loc[mask, 'volume'].describe().to_dict()
    print(f"\n{nombre}: {mask.sum()}/{total} ({pct:.1f}%)")
    print(f"  vol: min={vol_dist['min']:.0f} | p25={vol_dist['25%']:.0f} | p50={vol_dist['50%']:.0f} | p75={vol_dist['75%']:.0f} | max={vol_dist['max']:.0f}")
```
